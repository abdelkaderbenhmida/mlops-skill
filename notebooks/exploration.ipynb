# Customer Churn Exploration

Quick exploratory analysis of the synthetic telecom churn dataset.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/raw/dataset.csv')
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df['churn'].value_counts(normalize=True)
```

```python
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
numeric_cols = ['age', 'tenure_months', 'monthly_charges', 'total_charges', 'num_services', 'support_tickets']
for i, col in enumerate(numeric_cols):
    ax = axes[i // 3, i % 3]
    df[col].hist(bins=30, ax=ax, alpha=0.7)
    ax.set_title(col)
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
cat_cols = ['gender', 'region', 'contract_type', 'payment_method']
for i, col in enumerate(cat_cols):
    ax = axes[i // 2, i % 2]
    df.groupby(col)['churn'].mean().plot(kind='bar', ax=ax)
    ax.set_title(f'Churn rate by {col}')
    ax.set_ylabel('Churn rate')
plt.tight_layout()
plt.show()
```

```python
# Correlation with target
numeric_df = df.select_dtypes(include=[np.number])
corr = numeric_df.corr()['churn'].sort_values(ascending=False)
corr
```
